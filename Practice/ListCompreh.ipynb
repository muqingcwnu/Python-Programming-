List Comprehension :
List comprehension in Python is a concise way to create lists using a single line of code. It combines a loop, optional conditions, and an expression to generate list elements, often replacing longer for loops. The syntax is:

[expression for item in iterable if condition]

expression: The value or operation applied to each item (e.g., x*2).
item: The variable representing each element in the iterable.
iterable: A sequence like a list, tuple, or range.
condition (optional): A filter to include only certain items (e.g., if x > 0).

```python
square=[x**2 for x in range(10)]
print(square)
```

```python
evens=[x for x in range(10) if x%2==0]
print(evens)
```

```python
pairs = [(x, y) for x in [1,2] for y in [3,4]]
print(pairs)  # Output: [(1, 3), (1, 4), (2, 3), (2, 4)]
```

```python
pair=[(x,y) for x in [4,5] for y in [5,4]]
pair
pair.sort()
```
