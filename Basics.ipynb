What is Recursion?
Recursion is when a function calls itself until it doesn’t—meaning, it keeps repeating a smaller version of the problem until it hits a base case (the condition that stops the loop).



Let’s say we want to calculate factorial of a number (e.g., 5! = 5×4×3×2×1 = 120)

```python
def factorial(n):
    if n<0:
        return "Undefind for neagative numbers"
    if n==0:
        return 1
    else:
       return n*factorial(n-1)
print(factorial(5))
```

Fibonacci


```python
def fibo(n):
    if n<=1:
        return n
    else:
        return fibo(n-1)+fibo(n-2)
print(fibo(10))
```

```python
def fibo_sum(n):
    total=0
    for i in range (n):
        total+=fibo(i)
    return total
print(fibo_sum(10))
```

Reverse a string "Parvez"

```python
def reverse_string(s):
    if len(s)==0:
        return s
    else :
        return reverse_string(s[1:])+s[0]
print(reverse_string("Parvez"))
```

Reverse a list

```python
def reverse_list(lst):
    if len(lst)==0:
        return lst
    else :
        return [lst[-1]]+ reverse_list(lst[:-1])
print(reverse_list([1, 2, 3, 4, 5]))
```

2. Create a Person class with:
Attributes: name, birth_date (format: YYYY-MM-DD)

Method: calculate_age() that returns current age

Method: is_minor() that returns True if age < 18, else False

```python
from datetime import datetime

class Person:
    def __init__(self, name, birth_date):
        self.name = name
        self.birth_date = datetime.strptime(birth_date, "%Y-%m-%d")

    def calculate_age(self):
        today = datetime.today()
        age = today.year - self.birth_date.year - (
            (today.month, today.day) < (self.birth_date.month, self.birth_date.day)
        )
        return age

    def is_minor(self):
        return self.calculate_age() < 18


p = Person("PARVEZ", "2010-06-18")
print(p.name)
print(p.calculate_age())    
print(p.is_minor())         
```
