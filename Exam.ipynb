Recursion

```python
def fibnacci(n):
    if n<=1:
        return n
    else:
        return fibonacci(n-1) + fibonacci(n-2)
    
print(fibonacci(10))
```

```python
def factorial(n):
    if n==0 or n==1:
        return 1
    else:
        return n* factorial(n-1)
print(factorial(5))
```

```python
def reverse_string(s):
    if len(s)==0:
        return s
    else:
        return reverse_string(s[1:]) + s[0]
print(reverse_string("hello"))
```

```python
def find_max(lst):
    if len(lst)==1:
        return lst[0]
    else :
        return max(lst[0],find_max(lst[1:]))
print(find_max([1, 2, 3, 4, 5]))
```

```python
def find_max(lst):
    if len(lst)==1:
        return lst[0]
    else :
        return min(lst[0],find_max(lst[1:]))
print(find_max([1, 2, 3, 4, 5]))
```

```python
def is_prime(n):
    if n<=1:
        return False 
    for i in range(2,int(n**0.5)+1):
        if n%i==0:
            return False
    return True
for i in range(1,101):
    if is_prime(i):
        print(i, end=" ")
```
